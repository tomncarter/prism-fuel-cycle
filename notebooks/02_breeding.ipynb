# 02 — Blanket Breeding
Pu-239 mass accumulation in internal and radial blankets over time.
Capture rate proxy: absorption − fission from `blanket_breeding` tally.

```python
import pandas as pd
import matplotlib.pyplot as plt

inv = pd.read_parquet('../results/inventories.parquet')
tallies = pd.read_parquet('../results/tallies.parquet')

# Pu-239 mass in blanket materials
pu239 = inv[(inv['nuclide'] == 'Pu239') & (inv['material'].str.contains('blanket'))]
pu239.head()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for tag, grp in pu239.groupby(['run_tag', 'material']):
    run, mat = tag
    label = f'{run} | {mat}'
    axes[0].plot(grp['time_days'], grp['mass_g'] / 1e3, label=label, linewidth=1.5)

axes[0].set_xlabel('Time (days)')
axes[0].set_ylabel('Pu-239 mass (kg)')
axes[0].set_title('Pu-239 buildup in blankets')
axes[0].legend(fontsize=7)
axes[0].grid(True, alpha=0.3)

# Capture proxy from tally: absorption - fission
bb = tallies[tallies['tally'] == 'blanket_breeding'].copy()
# TODO: pivot score columns and compute capture = absorption - fission
axes[1].set_title('Blanket capture rate (placeholder)')
axes[1].set_xlabel('Time (days)')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('../results/02_breeding.png', dpi=150)
plt.show()
```
