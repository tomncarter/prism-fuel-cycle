# 03 — Flux Spectra
Energy-resolved flux in driver fuel and blankets at end of cycle.
Tracks flux hardening / softening over burnup.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

spec = pd.read_parquet('../results/spectra.parquet')
spec.head()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

for ax, tname, title in [
    (axes[0], 'flux_spectrum_fuel',    'Driver fuel flux spectrum'),
    (axes[1], 'flux_spectrum_blanket', 'Blanket flux spectrum'),
]:
    sub = spec[(spec['tally'] == tname)]
    for tag, grp in sub.groupby('run_tag'):
        # Energy group midpoints (eV)
        e_mid = np.sqrt(grp['energy low [eV]'] * grp['energy high [eV]'])
        ax.loglog(e_mid, grp['mean'], label=tag, linewidth=1.5)
    ax.set_xlabel('Energy (eV)')
    ax.set_ylabel('Flux (arb.)')
    ax.set_title(title)
    ax.legend(fontsize=7)
    ax.grid(True, which='both', alpha=0.3)

plt.tight_layout()
plt.savefig('../results/03_flux_spectra.png', dpi=150)
plt.show()
```
