# 01 — Criticality vs Time
k-effective evolution over the depletion cycle for all feed permutations.

```python
import pandas as pd
import matplotlib.pyplot as plt

keff = pd.read_parquet('../results/keff.parquet')
keff.head()
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

for tag, grp in keff.groupby('run_tag'):
    ax.errorbar(grp['time_days'], grp['keff'], yerr=grp['keff_unc'],
                label=tag, capsize=2, linewidth=1.5)

ax.axhline(1.0, color='k', linestyle='--', linewidth=0.8, label='Critical')
ax.set_xlabel('Time (days)')
ax.set_ylabel('k-effective')
ax.set_title('k-eff vs Burnup — all feed permutations')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('../results/01_criticality.png', dpi=150)
plt.show()
```
