# 04 — Neutron Leakage
Radial vs axial leakage current at the vessel boundary.
Spectrum of escaping neutrons from radial surface.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

tallies = pd.read_parquet('../results/tallies.parquet')
spec    = pd.read_parquet('../results/spectra.parquet')

# Leakage summary at final step
leak = tallies[tallies['tally'].str.startswith('leakage_')]
leak.head()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Bar chart: radial vs axial top vs axial bottom leakage per permutation
tally_names = ['leakage_radial_total', 'leakage_axial_top', 'leakage_axial_bot']
labels      = ['Radial', 'Axial top', 'Axial bot']

# TODO: pivot and plot once real data is available
axes[0].set_title('Leakage by direction (placeholder)')
axes[0].set_xlabel('Permutation')
axes[0].set_ylabel('Neutron current (arb.)')
axes[0].grid(True, alpha=0.3)

# Spectrum of radial leakage
rad_spec = spec[spec['tally'] == 'leakage_radial_spectrum']
for tag, grp in rad_spec.groupby('run_tag'):
    e_mid = np.sqrt(grp['energy low [eV]'] * grp['energy high [eV]'])
    axes[1].loglog(e_mid, grp['mean'], label=tag, linewidth=1.5)
axes[1].set_xlabel('Energy (eV)')
axes[1].set_ylabel('Leakage current (arb.)')
axes[1].set_title('Radial leakage spectrum')
axes[1].legend(fontsize=7)
axes[1].grid(True, which='both', alpha=0.3)

plt.tight_layout()
plt.savefig('../results/04_leakage.png', dpi=150)
plt.show()
```
